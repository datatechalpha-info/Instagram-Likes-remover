```python
import webbrowser
import pyautogui as p
import time
import pandas as pd
```

```python
# p.position()
unlike_btn = (1130,772)
for link in df1['links']:
    webbrowser.open(link)
    time.sleep(5)  # wait for the page to load
    p.click(unlike_btn)
    time.sleep(2)  # wait before closing the tab
    p.hotkey('ctrl', 'w')  # close the tab
    #also remove those links from the df1
    df1 = df1[df1 != link]
```
