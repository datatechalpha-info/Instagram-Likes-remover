```python
import webbrowser
import pyautogui as p
import time
import pandas as pd
```

```python
df1 = pd.read_csv('likes/posts_links.csv')
df1.head()
```

```python

for link in df1['links']:
    webbrowser.open(link)
    time.sleep(5)  # wait for the page to load
    try:
        unlike_btn = p.locateOnScreen("unlike.png", confidence=0.8)
        print(unlike_btn)
        if unlike_btn:
            p.click(unlike_btn)
            time.sleep(2)
            print(f"Unliked: {link}")
        else:
            print(f"Unlike button not found for: {link}")
    except p.ImageNotFoundException:
        print(f"Unlike button not found for: {link} (exception)")
    p.hotkey('ctrl', 'w')  # close the tab
    df1 = df1[df1['links'] != link]
```

```python
df1['visited'] = ''

for i,link in enumerate(df1['links']):
    df1.loc[i, 'visited'] = link
```
